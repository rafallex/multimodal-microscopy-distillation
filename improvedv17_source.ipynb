# Multimodal Cancer Classification Challenge 2026 — v17

**v11 LB = 0.6479 (high water mark). v15 LB = 0.572. v16 LB = 0.503.** Every "improvement" since v11 has made the LB worse. The OOD shift is severe enough that within-train methodology cannot bridge it.

## v17 strategy: undo everything since v11, then add the one trick that uses test data

**Reverted from v15/v16:**
- No CoMIR SSL pretraining
- No LOPO cross-validation (back to 3-fold stratified group)
- No aux NT-Xent alignment loss
- No EMA of weights
- No multi-snapshot ensembling
- No heavy stain aug
- No frozen projection heads

**Architecture & training: exact v11 reproduction.** Two-branch ResNet-18, 3-fold × 2 seeds + 1 full-data, AdamW + OneCycleLR @ LR=3e-4, mixup α=0.1, dropout 0.3, BCE + pos_weight, patient-balanced sampler, paired D4 aug, 8-way D4 TTA.

**The ONE new thing — pseudo-labels from v16:**
- Read v16's `submission_v16_combo.csv` (already produced by the previous Kaggle commit, uploaded as a Kaggle dataset)
- For each test cell with `p > 0.9`: pseudo-label as cancer (`Diagnosis = 1`)
- For each test cell with `p < 0.1`: pseudo-label as healthy (`Diagnosis = 0`)
- Drop the uncertain middle (`0.1 ≤ p ≤ 0.9`)
- Add the surviving cells to the train set, treating them as additional patients

**Why this should work:** the test set is OOD relative to train (which is why every within-train method has failed). The only way to inject test-distribution signal into training is to use the test data itself. Pseudo-labeling at confident tails is robust because v16's worst confusions are in the middle (mean prediction ≈ 0.5 on truly OOD patients); the tails are where v16 is least uncertain and therefore most likely correct.

## Compute budget on T4

| Stage | Time |
|---|---|
| JPEG cache (one-time) | ~45 min |
| 3-fold × 2 seeds supervised (10 epochs each, early stop on val_auc) | ~75 min |
| Full-data model | ~15 min |
| 8-way D4 TTA inference, 7 ckpts | ~30 min |
| **Total** | **~2h 45min** (well under 9h limit) |

```python
PAT_RE = re.compile(r"^pat_(\d+)_image_\d+\.jpg$")

def parse_patient_id(filename):
    m = PAT_RE.match(Path(filename).name)
    return int(m.group(1)) if m else None

def load_train_df(path):
    df = pd.read_csv(path); df.columns = [c.strip() for c in df.columns]
    df["patient_id"] = df["Name"].map(parse_patient_id).astype(int)
    df["split"] = "train"
    df["sample_weight"] = 1.0
    return df

def load_test_df(path):
    df = pd.read_csv(path); df.columns = [c.strip() for c in df.columns]
    return df

def cache_split(names, bf_dir, fl_dir, label=""):
    bf_dir, fl_dir = Path(bf_dir), Path(fl_dir)
    bf, fl = {}, {}
    t0 = time.time()
    for i, n in enumerate(names):
        with open(bf_dir / n, "rb") as f: bf[n] = f.read()
        with open(fl_dir / n, "rb") as f: fl[n] = f.read()
        if (i + 1) % 20000 == 0:
            print(f"  [{label}] cached {i+1}/{len(names)} in {time.time()-t0:.1f}s")
    print(f"  [{label}] cached {len(names)} in {time.time()-t0:.1f}s")
    return bf, fl

class CachedCellDataset(Dataset):
    """Reads BF + FL from RAM caches. Returns sample_weight so pseudo-labeled
    cells can be down-weighted in BCE."""
    def __init__(self, df, bf_cache_map, fl_cache_map, bf_tf, fl_tf, paired_tf=None):
        self.df = df.reset_index(drop=True)
        # bf_cache_map / fl_cache_map: {"train": train_cache, "test": test_cache}
        self.bf_cache_map = bf_cache_map
        self.fl_cache_map = fl_cache_map
        self.bf_tf = bf_tf; self.fl_tf = fl_tf
        self.paired_tf = paired_tf
    def __len__(self): return len(self.df)
    @staticmethod
    def _decode(buf): return Image.open(io.BytesIO(buf)).convert("L")
    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        name = row["Name"]
        split = row.get("split", "train")
        bf = self.bf_tf(self._decode(self.bf_cache_map[split][name]))
        fl = self.fl_tf(self._decode(self.fl_cache_map[split][name]))
        if self.paired_tf is not None:
            bf, fl = self.paired_tf(bf, fl)
        label = int(row["Diagnosis"]) if "Diagnosis" in row else -1
        weight = float(row.get("sample_weight", 1.0))
        return {"bf": bf, "fl": fl, "label": label, "name": name, "weight": weight}
```

```python
def stratified_patient_kfold(df, n_splits=3, seed=1):
    """3-fold split on TRAIN patients only. Pseudo-labeled test patients are
    appended to every train fold (so all CV models see them) but never enter
    the validation set."""
    train_only = df[df["split"] == "train"].reset_index(drop=True)
    skgf = StratifiedGroupKFold(n_splits=n_splits, shuffle=True, random_state=seed)
    y = train_only["Diagnosis"].to_numpy(); groups = train_only["patient_id"].to_numpy()
    splits = list(skgf.split(train_only, y=y, groups=groups))
    for f, (_, va) in enumerate(splits):
        if len(np.unique(y[va])) < 2:
            raise ValueError(f"Fold {f} has only one class — try another seed.")
    # Map indices back to the full (train + pseudo) dataframe.
    train_orig_idx = df.index[df["split"] == "train"].to_numpy()
    pseudo_idx = df.index[df["split"] == "test"].to_numpy()
    out = []
    for tr, va in splits:
        tr_full = np.concatenate([train_orig_idx[tr], pseudo_idx])
        va_full = train_orig_idx[va]
        out.append((tr_full, va_full))
    return out

def summarize_split(df, tr, va):
    trd, vad = df.iloc[tr], df.iloc[va]
    pseudo_in_train = int((trd["split"] == "test").sum())
    return (f"train: {len(tr):>6} cells ({pseudo_in_train} pseudo), {trd['patient_id'].nunique():>2}p, "
            f"pos {trd['Diagnosis'].mean():.3f} | "
            f"val: {len(va):>6} cells, {vad['patient_id'].nunique():>2}p, "
            f"pos {vad['Diagnosis'].mean():.3f} | "
            f"val pats: {sorted(vad['patient_id'].unique().tolist())}")

class PatientBalancedSampler(Sampler):
    """Each batch contains cells from `patients_per_batch` distinct patients.
    Pseudo-labeled test patients participate in sampling identically."""
    def __init__(self, df, batch_size, patients_per_batch=4, seed=0):
        assert batch_size % patients_per_batch == 0
        self.df = df.reset_index(drop=True)
        self.batch_size = batch_size
        self.per_pat = batch_size // patients_per_batch
        self.patients_per_batch = patients_per_batch
        self.rng = np.random.default_rng(seed)
        self.by_pat = {p: np.array(g.index.tolist())
                       for p, g in self.df.groupby("patient_id")}
        self.patients = list(self.by_pat.keys())
        self.epoch_len = len(self.df) // batch_size * batch_size
    def __len__(self): return self.epoch_len
    def __iter__(self):
        out = []
        for _ in range(self.epoch_len // self.batch_size):
            pats = self.rng.choice(self.patients,
                                   size=min(self.patients_per_batch, len(self.patients)),
                                   replace=False)
            for p in pats:
                idxs = self.by_pat[p]
                out.extend(self.rng.choice(idxs, size=self.per_pat,
                                           replace=len(idxs) < self.per_pat).tolist())
        return iter(out)
```

```python
def _to_tensor_norm(mean, std):
    def fn(img):
        t = TF.to_tensor(img)
        return TF.normalize(t, [mean], [std])
    return fn

to_tensor_bf = _to_tensor_norm(BF_MEAN, BF_STD)
to_tensor_fl = _to_tensor_norm(FL_MEAN, FL_STD)

class PairedGeoAug:
    """Same geometric transform on BF and FL: D4 + mild rotation."""
    def __init__(self, p_hflip=0.5, p_vflip=0.5, rot90=True, max_rot=10.0):
        self.p_hflip = p_hflip; self.p_vflip = p_vflip
        self.rot90 = rot90; self.max_rot = max_rot
    def __call__(self, bf, fl):
        if self.rot90:
            k = random.randint(0, 3)
            if k:
                bf = torch.rot90(bf, k, dims=(-2, -1))
                fl = torch.rot90(fl, k, dims=(-2, -1))
        if random.random() < self.p_hflip: bf, fl = TF.hflip(bf), TF.hflip(fl)
        if random.random() < self.p_vflip: bf, fl = TF.vflip(bf), TF.vflip(fl)
        if self.max_rot > 0:
            a = random.uniform(-self.max_rot, self.max_rot)
            bf, fl = TF.rotate(bf, a), TF.rotate(fl, a)
        return bf, fl

def train_modality_transform(modality):
    """Mild ColorJitter (v11 setting, NOT v15/v16 heavy aug)."""
    norm = to_tensor_bf if modality == "bf" else to_tensor_fl
    return T.Compose([T.ColorJitter(brightness=0.2, contrast=0.2), norm])

def eval_modality_transform(modality):
    return to_tensor_bf if modality == "bf" else to_tensor_fl
```

```python
def _make_resnet18_branch(pretrained=True):
    weights = "DEFAULT" if pretrained else None
    net = models.resnet18(weights=weights)
    w = net.conv1.weight.data
    new_conv = nn.Conv2d(1, 64, kernel_size=7, stride=2, padding=3, bias=False)
    if pretrained:
        new_conv.weight.data = w.mean(dim=1, keepdim=True)
    net.conv1 = new_conv
    fd = net.fc.in_features; net.fc = nn.Identity()
    return net, fd  # 512

class MultimodalClassifier(nn.Module):
    """v11 architecture exactly: two ResNet-18 branches + concat fusion head."""
    def __init__(self, pretrained=True, dropout=DROPOUT):
        super().__init__()
        self.bf_branch, fd = _make_resnet18_branch(pretrained)
        self.fl_branch, _  = _make_resnet18_branch(pretrained)
        self.head = nn.Sequential(
            nn.Linear(fd * 2, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Dropout(dropout),
            nn.Linear(256, 1),
        )
    def forward(self, bf, fl):
        feat = torch.cat([self.bf_branch(bf), self.fl_branch(fl)], dim=1)
        return self.head(feat).squeeze(-1)

with torch.no_grad():
    _m = MultimodalClassifier(pretrained=False).cpu()
    _x = torch.zeros(2, 1, 128, 128)
    print("Output shape:", _m(_x, _x).shape,
          "  params:", sum(p.numel() for p in _m.parameters()) // 1_000_000, "M")
    del _m, _x
```

```python
# === Build pseudo-labeled dataset from v16's predictions ===
# Filter v16's test predictions to high-confidence tails. The middle (uncertain)
# is exactly where v16 is wrong on OOD patients, so we drop it.

if PSEUDO_SRC_PATH is not None:
    v16_preds = pd.read_csv(PSEUDO_SRC_PATH)
    assert set(v16_preds.columns) == {"Name", "Diagnosis"}, \
        f"unexpected columns in {PSEUDO_SRC_PATH}: {v16_preds.columns.tolist()}"
    v16_preds = v16_preds.rename(columns={"Diagnosis": "v16_pred"})
    print(f"Loaded {len(v16_preds)} v16 predictions")
    print(f"  v16 pred distribution: min={v16_preds['v16_pred'].min():.3f} "
          f"max={v16_preds['v16_pred'].max():.3f} mean={v16_preds['v16_pred'].mean():.3f}")

    # Per-test-patient confidence stats — useful for diagnostics.
    test_with_pat = v16_preds.copy()
    test_with_pat["patient_id"] = test_with_pat["Name"].map(parse_patient_id)
    pat_stats = (test_with_pat.groupby("patient_id")["v16_pred"]
                 .agg(["mean", "std", "count"]).sort_values("mean"))
    print("\nv16 per-test-patient prediction stats:")
    print(pat_stats.to_string())

    # Apply confidence threshold and build pseudo-labeled rows.
    pos = v16_preds[v16_preds["v16_pred"] > PSEUDO_THRESHOLD_POS].copy()
    neg = v16_preds[v16_preds["v16_pred"] < PSEUDO_THRESHOLD_NEG].copy()
    pos["Diagnosis"] = 1
    neg["Diagnosis"] = 0
    pseudo = pd.concat([pos, neg], ignore_index=True)
    pseudo["patient_id"] = pseudo["Name"].map(parse_patient_id).astype(int)
    pseudo["split"] = "test"
    pseudo["sample_weight"] = PSEUDO_WEIGHT
    pseudo = pseudo[["Name", "Diagnosis", "patient_id", "split", "sample_weight"]]

    print(f"\nPseudo-labels @ thresholds [{PSEUDO_THRESHOLD_NEG}, {PSEUDO_THRESHOLD_POS}]:")
    print(f"  positives (pseudo=1): {len(pos)}  from {pos['Name'].map(parse_patient_id).nunique()} test patients")
    print(f"  negatives (pseudo=0): {len(neg)}  from {neg['Name'].map(parse_patient_id).nunique()} test patients")
    print(f"  total pseudo-labeled cells: {len(pseudo)}  ({len(pseudo)/len(v16_preds)*100:.1f}% of test)")
    print(f"  sample_weight: {PSEUDO_WEIGHT}")

    # Verify no Name overlap with train (sanity check).
    train_names = set(df_train["Name"])
    overlap = set(pseudo["Name"]) & train_names
    assert not overlap, f"pseudo-labels overlap train names: {list(overlap)[:5]}"

    # Verify patient IDs don't collide with train patients.
    train_pats = set(df_train["patient_id"])
    pseudo_pats = set(pseudo["patient_id"])
    pat_overlap = train_pats & pseudo_pats
    assert not pat_overlap, f"pseudo patient IDs overlap train: {pat_overlap}"

    df_combined = pd.concat([df_train, pseudo], ignore_index=True)
    print(f"\nCombined dataset: {len(df_combined)} cells, {df_combined['patient_id'].nunique()} unique patients")
    print(f"  train patients: {sorted(train_pats)} ({len(train_pats)})")
    print(f"  pseudo patients (from test): {sorted(pseudo_pats)} ({len(pseudo_pats)})")
else:
    df_combined = df_train.copy()
    print("No pseudo-label source — using train only (pure v11 reproduction).")
```

```python
def mixup_batch(bf, fl, y, w, alpha=0.1):
    """Mixup on inputs + labels + sample weights jointly."""
    lam = float(np.random.beta(alpha, alpha))
    idx = torch.randperm(bf.size(0), device=bf.device)
    return (lam * bf + (1 - lam) * bf[idx],
            lam * fl + (1 - lam) * fl[idx],
            lam * y  + (1 - lam) * y[idx],
            lam * w  + (1 - lam) * w[idx])

def run_epoch(model, loader, optimizer, scaler, criterion_unreduced, train,
              mixup_alpha=0.0, grad_clip=0.0, sched=None, log_every=0):
    """v11 epoch loop, with per-sample weighting so pseudo-labels can be down-weighted.
    `criterion_unreduced` must be BCEWithLogitsLoss(reduction='none')."""
    model.train(train)
    losses, hard_ys, ps = [], [], []
    t_last = time.time()
    for i, batch in enumerate(loader):
        bf = batch["bf"].to(DEVICE, non_blocking=True)
        fl = batch["fl"].to(DEVICE, non_blocking=True)
        y  = batch["label"].float().to(DEVICE, non_blocking=True)
        w  = batch["weight"].float().to(DEVICE, non_blocking=True)
        hard_ys.append(batch["label"].numpy())

        if train and mixup_alpha > 0:
            bf, fl, y, w = mixup_batch(bf, fl, y, w, mixup_alpha)

        with torch.amp.autocast("cuda", enabled=scaler is not None):
            logits = model(bf, fl)
            per_sample = criterion_unreduced(logits, y)
            loss = (per_sample * w).sum() / w.sum().clamp_min(1e-6)

        if train:
            optimizer.zero_grad(set_to_none=True)
            if scaler is not None:
                scaler.scale(loss).backward()
                if grad_clip > 0:
                    scaler.unscale_(optimizer)
                    nn.utils.clip_grad_norm_(model.parameters(), grad_clip)
                old_scale = scaler.get_scale()
                scaler.step(optimizer); scaler.update()
                if sched is not None and scaler.get_scale() >= old_scale:
                    sched.step()
            else:
                loss.backward()
                if grad_clip > 0: nn.utils.clip_grad_norm_(model.parameters(), grad_clip)
                optimizer.step()
                if sched is not None: sched.step()

        losses.append(loss.item())
        ps.append(torch.sigmoid(logits).detach().float().cpu().numpy())
        if log_every and (i + 1) % log_every == 0:
            dt = time.time() - t_last; t_last = time.time()
            print(f"    step {i+1}/{len(loader)} | {dt:.1f}s | loss {float(np.mean(losses[-log_every:])):.4f}")
    hard_ys = np.concatenate(hard_ys); ps = np.concatenate(ps)
    auc = roc_auc_score(hard_ys, ps) if len(np.unique(hard_ys)) > 1 else float("nan")
    return float(np.mean(losses)), auc, hard_ys, ps

def train_one_model(train_df, val_df, ckpt_path, oof_path, hist_path,
                   epochs, seed, sampler_kind="patient", track_oof=True):
    """Train one model on `train_df`, validate on `val_df` (can be None for full-data)."""
    seed_everything(seed)

    train_ds = CachedCellDataset(train_df, bf_cache_map, fl_cache_map,
                                 train_modality_transform("bf"),
                                 train_modality_transform("fl"),
                                 paired_tf=PairedGeoAug())
    if sampler_kind == "patient":
        sampler = PatientBalancedSampler(train_df, batch_size=BATCH_SIZE,
                                         patients_per_batch=PATIENTS_PER_BATCH, seed=seed)
        train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, sampler=sampler,
                                  num_workers=NUM_WORKERS, pin_memory=True, drop_last=True)
    else:
        train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,
                                  num_workers=NUM_WORKERS, pin_memory=True, drop_last=True)
    val_loader = None
    if val_df is not None:
        val_ds = CachedCellDataset(val_df, bf_cache_map, fl_cache_map,
                                   eval_modality_transform("bf"),
                                   eval_modality_transform("fl"))
        val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE * 2, shuffle=False,
                                num_workers=NUM_WORKERS, pin_memory=True)

    model = MultimodalClassifier(pretrained=True, dropout=DROPOUT).to(DEVICE)
    # pos_weight based on the REAL train labels only (ignore pseudo so the
    # weighting reflects the true class prior).
    real_train = train_df[train_df["split"] == "train"]
    pos = (real_train["Diagnosis"] == 1).sum()
    neg = (real_train["Diagnosis"] == 0).sum()
    pos_weight = torch.tensor(neg / max(pos, 1), device=DEVICE)
    print(f"  pos_weight={pos_weight.item():.3f}  seed={seed}  epochs={epochs}  "
          f"train_n={len(train_df)} (real={len(real_train)}, pseudo={len(train_df)-len(real_train)})")
    # Use 'none' reduction so we can apply per-sample weights manually.
    criterion_unreduced = nn.BCEWithLogitsLoss(pos_weight=pos_weight, reduction="none")

    optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
    sched = torch.optim.lr_scheduler.OneCycleLR(
        optimizer, max_lr=LR, steps_per_epoch=len(train_loader),
        epochs=epochs, pct_start=0.1)
    scaler = torch.amp.GradScaler("cuda") if DEVICE == "cuda" else None

    history, best_auc, best_ep, no_improve = [], -1.0, 0, 0
    for ep in range(epochs):
        t0 = time.time()
        tr_loss, tr_auc, _, _ = run_epoch(
            model, train_loader, optimizer, scaler, criterion_unreduced, True,
            mixup_alpha=MIXUP_ALPHA, grad_clip=GRAD_CLIP, sched=sched, log_every=200)
        va_loss = va_auc = float("nan"); vy = vp = None
        if val_loader is not None:
            with torch.no_grad():
                va_loss, va_auc, vy, vp = run_epoch(
                    model, val_loader, None, None, criterion_unreduced, False)
        dt = time.time() - t0
        print(f"  ep {ep:>2d} | tr_loss {tr_loss:.4f} tr_auc {tr_auc:.4f} "
              f"| va_loss {va_loss:.4f} va_auc {va_auc:.4f} | {dt:.1f}s")
        history.append({"epoch": ep, "tr_loss": tr_loss, "tr_auc": tr_auc,
                        "va_loss": va_loss, "va_auc": va_auc, "time": dt})
        save_now = False
        if val_loader is not None:
            if va_auc > best_auc:
                best_auc, best_ep, no_improve = va_auc, ep, 0; save_now = True
            else:
                no_improve += 1
        else:
            save_now = True; best_ep = ep
        if save_now:
            torch.save({"model": model.state_dict(), "epoch": ep,
                        "val_auc": va_auc if val_loader is not None else None,
                        "args": {"dropout": DROPOUT}}, ckpt_path)
            if track_oof and val_loader is not None:
                pd.DataFrame({"Name": val_df["Name"].values,
                              "patient_id": val_df["patient_id"].values,
                              "y_true": vy, "y_pred": vp}).to_csv(oof_path, index=False)
        if val_loader is not None and no_improve >= PATIENCE:
            print(f"  Early stopping at epoch {ep}"); break
    with open(hist_path, "w") as f:
        json.dump({"history": history, "best_auc": best_auc, "best_ep": best_ep}, f, indent=2)
    del model, optimizer, sched, scaler, train_loader
    if val_loader is not None: del val_loader
    gc.collect()
    if torch.cuda.is_available(): torch.cuda.empty_cache()
    return best_auc, best_ep
```

```python
def _d4(bf, fl):
    for k in range(4):
        bfr = torch.rot90(bf, k, dims=(-2, -1))
        flr = torch.rot90(fl, k, dims=(-2, -1))
        yield bfr, flr
        yield TF.hflip(bfr), TF.hflip(flr)

def load_model_from_ckpt(ckpt_path):
    state = torch.load(ckpt_path, map_location=DEVICE, weights_only=False)
    args = state.get("args", {})
    dropout = float(args.get("dropout", DROPOUT))
    model = MultimodalClassifier(pretrained=False, dropout=dropout).to(DEVICE)
    model.load_state_dict(state["model"]); model.eval()
    return model

def predict_one_ckpt(ckpt_path, loader, tta=True):
    model = load_model_from_ckpt(ckpt_path)
    preds = []
    n_aug = 8 if tta else 1
    with torch.no_grad():
        for batch in loader:
            bf = batch["bf"].to(DEVICE, non_blocking=True)
            fl = batch["fl"].to(DEVICE, non_blocking=True)
            p = None
            for bf_t, fl_t in (_d4(bf, fl) if tta else [(bf, fl)]):
                with torch.amp.autocast("cuda", enabled=DEVICE == "cuda"):
                    pi = torch.sigmoid(model(bf_t, fl_t)).float()
                p = pi if p is None else p + pi
            preds.append((p / n_aug).cpu().numpy())
    del model
    if torch.cuda.is_available(): torch.cuda.empty_cache()
    return np.concatenate(preds)

# All test cells get "split=test" so the CachedCellDataset reads from the test cache.
df_test_for_pred = df_test.copy()
df_test_for_pred["patient_id"] = df_test_for_pred["Name"].map(parse_patient_id)
df_test_for_pred["split"] = "test"
df_test_for_pred["sample_weight"] = 1.0
test_ds = CachedCellDataset(df_test_for_pred, bf_cache_map, fl_cache_map,
                            eval_modality_transform("bf"),
                            eval_modality_transform("fl"))
test_loader = DataLoader(test_ds, batch_size=256, shuffle=False,
                         num_workers=NUM_WORKERS, pin_memory=True)

ckpts = sorted(glob.glob(str(OUT_DIR / "fold*_best.pt")))
if TRAIN_FULL_DATA_MODEL and (OUT_DIR / "fulldata_best.pt").exists():
    ckpts.append(str(OUT_DIR / "fulldata_best.pt"))
print(f"Ensembling {len(ckpts)} models:")
for c in ckpts: print("  -", Path(c).name)

all_preds = []
for c in ckpts:
    t0 = time.time()
    all_preds.append(predict_one_ckpt(c, test_loader, tta=True))
    print(f"  {Path(c).name} done in {time.time()-t0:.1f}s")
preds = np.mean(all_preds, axis=0)

sub = pd.DataFrame({"Name": df_test["Name"].values, "Diagnosis": preds})
sub.to_csv("/kaggle/working/submission.csv", index=False)
print(f"\nWrote submission.csv  (mean pred = {preds.mean():.3f}, "
      f"min {preds.min():.3f}, max {preds.max():.3f})")
print(sub.head())
!wc -l /kaggle/working/submission.csv
```
